Download an image and then download its saliency maps for all models and labels

```python
import cv2
import matplotlib.pyplot as plt
import pandas as pd
import os

import sys
sys.path.append('.')
from config import PROJECT_DIR, DATASET_DIR, DATASET_ZIP
```

```python
os.system(f'unzip -q {DATASET_ZIP} -d {PROJECT_DIR}')
```

```python
IMG_DIR = f'{DATASET_DIR}/images_1024'
```

```python
df = pd.read_csv(f'{PROJECT_DIR}/predictions_by_image.csv')
files = set([f.replace('.png', '') for f in os.listdir(IMG_DIR)])

def save_images(label, n, output_dir=f'{PROJECT_DIR}/x-rays'):
    """
    label: 'healthy', 'aneurysm', 'cardiomegaly', 'both'
    n: number of images to download
    output_dir: directory where all the images will be stored, default: {PROJECT_DIR/x-rays}
    """
    os.makedirs(output_dir, exist_ok=True)
    df_model = df[df['model'] == 'DN_1024'].copy()

    if label == 'healthy':
        df_fil = df_model[(df_model['label_aneurysm'] == 0) & (df_model['label_cardiomegaly'] == 0)]
    elif label == 'aneurysm':
        df_fil = df_model[(df_model['label_aneurysm'] == 1) & (df_model['label_cardiomegaly'] == 0)]
    elif label == 'cardiomegaly':
        df_fil = df_model[(df_model['label_aneurysm'] == 0) & (df_model['label_cardiomegaly'] == 1)]
    elif label == 'both':
        df_fil = df_model[(df_model['label_aneurysm'] == 1) & (df_model['label_cardiomegaly'] == 1)]

    df_fil = df_fil[df_fil['image_id'].isin(files)]

    for i in range(min(n, len(df_fil))):
        image_id = df_fil.iloc[i]['image_id']
        img = cv2.imread(os.path.join(IMG_DIR, f'{image_id}.png'), cv2.IMREAD_GRAYSCALE)
        fig, ax = plt.subplots(figsize=(6, 6))
        ax.imshow(img, cmap='gray')
        ax.axis('off')
        plt.tight_layout(pad=0)
        path = os.path.join(output_dir, f'{label}_{i+1}.png')
        plt.savefig(path, dpi=150, bbox_inches='tight', pad_inches=0)
        plt.close()
        print(f'Saved: {path}')

# Use:
# save_images('healthy', 3, f'{PROJECT_DIR}/healthy_x_rays')
# save_images('cardiomegaly', 8)
# save_images('both', 2, f'{PROJECT_DIR}/aortic_enlargement_and_cardiomegaly_x_rays')
```
